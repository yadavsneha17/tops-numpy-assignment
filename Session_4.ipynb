```python
import numpy as np
```

```python
#Task 1
likes = np.array([100, 200, 300, 400, 500, 600, 700])
comments = np.array([10, 20, 30, 40, 50, 60, 70])
engagement = (likes + comments) / 2
print(engagement)
```

```python
# Task 2
prices = np.array([200, 300, 500, 400, 600])
discounts = np.array([20, 50, 100, 40, 60])
final_price = prices - discounts
print(final_price)
```

```python
# Task 3
scores = np.array([150, 190, 175, 200, 160])
result = scores > 180
print(result)
```

```python
# Task 4
paytm = np.array([1, 0, 1, 0, 1, 0])
phonepe = np.array([0, 1, 1, 0, 0, 1])
result = np.logical_or(paytm, phonepe)
print(result)
```

```python
# Task 5
steps = np.array([4000, 5000, 6000, 3000, 7000, 8000, 5000])
bonus_steps = steps + 500
total_steps = bonus_steps.sum()
print("Steps:", bonus_steps)
print("Total steps:", total_steps)
```
