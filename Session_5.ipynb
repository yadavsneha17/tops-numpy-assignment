```python
import numpy as np
```

```python
# Task 1
likes = np.array([100, 200, 300, 400, 500, 600, 700])
friend_likes = np.array([150, 180, 350, 300, 450, 650, 800])

print("Combined:", np.add(likes, friend_likes))
print("Difference:", np.subtract(likes, friend_likes))
```

```python
# Task 2
prices = np.array([200, 300, 500, 400, 600])
discounted = np.multiply(prices, 0.9)
print("Discounted prices:", discounted)
print("Final Bill:", np.sum(discounted))
```

```python
# Task 3
steps = np.array([
    4000, 5000, 6000, 7000, 4500,
    8000, 6500, 5500, 7200, 4800,
    5000, 6000, 7500, 8200, 4300,
    5200, 6100, 7000, 6800, 5900,
    4500, 5600, 6300, 7100, 8000,
    4900, 5800, 6700, 7300, 6200
])

print("Mean:", np.mean(steps))
print("Median:", np.median(steps))
print("Standard Deviation:", np.std(steps))
print("Maximum:", np.max(steps))
```

```python
# Task 4
ratings = np.random.uniform(1, 5, 10)
print("Ratings:", ratings)
print("Round:", np.round(ratings))
print("Floor:", np.floor(ratings))
print("Ceil:", np.ceil(ratings))
```

```python
# Task 5
skips = np.array([
    0, 1, 0, 1, 1, 0, 2, 1, 0, 1,
    2, 0, 1, 1, 0, 2, 1, 0, 1, 2
])
percentage = np.percentile(skips, 75)
print("75th Percentile:", percentage)
```
