```python
# Task 1
zomato = np.array([4.5, 4.0, 3.5, 4.2, 5.0])
swiggy = np.array([4.0, 3.8, 4.5, 4.1, 4.7])
result = np.concatenate((zomato, swiggy))
print(result)
```

```python
# Task 2
post1 = np.array([100, 200, 150, 300, 250, 180, 220])
post2 = np.array([150, 250, 200, 350, 300, 220, 280])
post3 = np.array([80, 120, 180, 200, 250, 300, 350])

result = np.vstack((post1, post2, post3))
print(result)
```

```python
# Task 4
messages = np.array([
    [10, 20, 15, 12, 18, 20, 25],
    [5, 8, 10, 7, 6, 9, 11],
    [30, 25, 28, 35, 32, 30, 40]
])

new_user = np.array([[15, 18, 20, 16, 22, 19, 25]])
messages = np.insert(messages, 3, new_user, axis=0)
least = np.argmin(messages.sum(axis=1))
messages = np.delete(messages, least, axis=0)
print(messages)
```
