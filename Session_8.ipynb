```python
import numpy as np
```

```python
#Task 1
teams = np.array(["CSK", "MI", "RCB", "MI", "CSK", "GT", "RCB"])
unique_teams = np.unique(teams)
print(unique_teams)
```

```python
# Task 2
ratings = np.array([4.5, np.nan, 4.0, np.nan, 3.5])
missing = np.isnan(ratings)
print("Missing ratings:", np.sum(missing))
```

```python
# Task 3
prices = np.array([50, 200, 500, 1200, 800, 1500])
result = np.clip(prices, 100, 1000)
print(result)
```

```python
#Task 4
views = np.array([1000, 2000, np.nan, 5000, np.inf])
max_value = np.nanmax(views[np.isfinite(views)])
views[np.isnan(views)] = 0
views[np.isinf(views)] = max_value
print(views)
```
